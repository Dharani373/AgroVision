```python
import joblib
from xgboost import XGBRegressor

# Load the saved model and preprocessor
loaded_model = XGBRegressor()
loaded_model.load_model("../models/yield_model.json")

loaded_preprocessor = joblib.load("../models/preprocessor.pkl")

print("Model loaded successfully!")
print("Preprocessor loaded successfully!")
```

```python
import pandas as pd

# Load the original dataset
df = pd.read_csv("../data/raw/Crop Yeild Data.csv")

# Remove extra spaces from text columns
for col in ["Crop", "Season", "State"]:
    df[col] = df[col].str.strip()

print("Dataset loaded:", df.shape)
```

```python
# Remove target and unused production column
X = df.drop(columns=["Yield", "Production"])
y = df["Yield"]

# Take one row as a test prediction
sample = X.iloc[[0]]

# Transform the row using the saved preprocessor
sample_encoded = loaded_preprocessor.transform(sample)

# Predict the yield
prediction = loaded_model.predict(sample_encoded)[0]

print("Input:")
print(sample.to_string(index=False))

print("\nActual Yield:", y.iloc[0])
print("Predicted Yield:", prediction)
```
